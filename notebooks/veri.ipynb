```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Örnek veri setini oluşturuyoruz (Gerçek veri ile değiştirebilirsin)
data = {
    "Episode": np.arange(1, 101),
    "Gerçek Değer": np.random.uniform(4, 6, 100),  # Gerçek ödüller (örnek veriler)
    "OIS Tahmini": np.random.uniform(3.5, 6.5, 100),  # Ordinary Importance Sampling tahminleri
    "WIS Tahmini": np.random.uniform(4, 6, 100),  # Weighted Importance Sampling tahminleri
    "PDIS Tahmini": np.random.uniform(4.2, 5.8, 100),  # Per-Decision Importance Sampling tahminleri
}

df = pd.DataFrame(data)

# MSE hesaplamaları
df["MSE_OIS"] = (df["Gerçek Değer"] - df["OIS Tahmini"])**2
df["MSE_WIS"] = (df["Gerçek Değer"] - df["WIS Tahmini"])**2
df["MSE_PDIS"] = (df["Gerçek Değer"] - df["PDIS Tahmini"])**2

# Ortalama MSE değerleri
mse_ois = df["MSE_OIS"].mean()
mse_wis = df["MSE_WIS"].mean()
mse_pdis = df["MSE_PDIS"].mean()

print(f"MSE (Ordinary Importance Sampling): {mse_ois:.4f}")
print(f"MSE (Weighted Importance Sampling): {mse_wis:.4f}")
print(f"MSE (Per-Decision Importance Sampling): {mse_pdis:.4f}")

# MSE grafiğini çizme
plt.figure(figsize=(8,5))
plt.plot(df["Episode"], df["MSE_OIS"], label="Ordinary Importance Sampling", color="green")
plt.plot(df["Episode"], df["MSE_WIS"], label="Weighted Importance Sampling", color="red")
plt.plot(df["Episode"], df["MSE_PDIS"], label="Per-Decision Importance Sampling", color="blue")

plt.xlabel("Episodes")
plt.ylabel("Mean Squared Error (MSE)")
plt.title("MSE Comparison of Importance Sampling Methods")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Episode sayısını artırarak kitapta belirtilen uzun vadeli karşılaştırmayı taklit ediyoruz
episodes = np.logspace(0, 4, num=200, dtype=int)  # Log ölçekli episode aralığı

# Gerçek değerler ve farklı yöntemler için tahminler (daha belirgin varyans farkları ile)
true_values = np.ones(len(episodes))  # Gerçek ödüller (tüm yöntemler için referans)
ois_estimates = true_values + np.random.normal(0, 0.3, len(episodes))  # OIS için hata
wis_estimates = true_values + np.random.normal(0, 0.2, len(episodes))  # WIS için hata
pdis_estimates = true_values + np.random.normal(0, 0.15, len(episodes))  # PDIS için hata

# MSE hesaplamaları (gerçek değere göre)
mse_ois = np.square(ois_estimates - true_values)
mse_wis = np.square(wis_estimates - true_values)
mse_pdis = np.square(pdis_estimates - true_values)

# Grafik oluşturma
fig, axs = plt.subplots(2, 1, figsize=(10, 10))

# İlk grafik: OIS vs WIS
axs[0].plot(episodes, mse_ois, label="Ordinary Importance Sampling", color="green")
axs[0].plot(episodes, mse_wis, label="Weighted Importance Sampling", color="red")
axs[0].set_xscale("log")
axs[0].set_xlabel("Episodes (log scale)")
axs[0].set_ylabel("Mean Squared Error (MSE)")
axs[0].set_title("MSE Comparison of Ordinary vs. Weighted Importance Sampling")
axs[0].legend()
axs[0].grid(True, which="both", linestyle="--", linewidth=0.5)

# İkinci grafik: WIS vs PDIS
axs[1].plot(episodes, mse_wis, label="Weighted Importance Sampling", color="red")
axs[1].plot(episodes, mse_pdis, label="Per-Decision Importance Sampling", color="blue")
axs[1].set_xscale("log")
axs[1].set_xlabel("Episodes (log scale)")
axs[1].set_ylabel("Mean Squared Error (MSE)")
axs[1].set_title("MSE Comparison of Weighted vs. Per-Decision Importance Sampling")
axs[1].legend()
axs[1].grid(True, which="both", linestyle="--", linewidth=0.5)

# Grafikleri göster
plt.tight_layout()
plt.show()
```
