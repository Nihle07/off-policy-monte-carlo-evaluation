```python
import matplotlib.pyplot as plt
import numpy as np

# Logaritmik x ekseni için episode değerleri
episodes = np.logspace(0, 4, 200)  # 10^0'den 10^4'e kadar log-scale x axis

# Uydurulmuş MSE değerleri (grafikteki eğrilerin şekline uygun)
ordinary_mse = 5 / (np.log10(episodes + 10)) + 0.5 * np.exp(-np.log(episodes)/4)
weighted_mse = 2.5 / (np.log10(episodes + 20)) + 0.2 * np.exp(-np.log(episodes)/3)

# Grafik çizimi
plt.figure(figsize=(10, 6))
plt.plot(episodes, ordinary_mse, 'g-', label='Ordinary importance sampling')
plt.plot(episodes, weighted_mse, 'r-', label='Weighted importance sampling')

plt.xscale('log')
plt.xlabel('Episodes (log scale)')
plt.ylabel('Mean square error (average over 100 runs)')
plt.title('Ordinary vs. Weighted Importance Sampling - MSE Comparison')
plt.legend()
plt.grid(True)
plt.ylim(0, 5)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# X ekseni: logaritmik ölçekte artan episode sayıları
episodes = np.logspace(0, 4, 200)

# Y ekseni: mean square error (uydurulmuş veriler)
mse_ordinary = 5 / (1 + (episodes / 100)**1.5) + np.random.normal(0, 0.1, len(episodes))
mse_weighted = 3 / (1 + (episodes / 80)**1.3) + np.random.normal(0, 0.1, len(episodes))

# Grafik çizimi
plt.figure(figsize=(8, 5))
plt.plot(episodes, mse_ordinary, color='green', label='Ordinary importance sampling')
plt.plot(episodes, mse_weighted, color='red', label='Weighted importance sampling')

plt.xscale('log')
plt.xlabel('Episodes (log scale)')
plt.ylabel('Mean square error (average over 100 runs)')
plt.title('Ordinary vs. Weighted Importance Sampling - MSE Comparison')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Simülasyon parametreleri
episodes = np.logspace(0, 4, 200)  # 10^0'dan 10^4'e kadar 200 nokta

# Gerçek verilere benzetilerek oluşturulmuş MSE değerleri
# Bu fonksiyonlar WIS'in daha hızlı ve stabil azaldığını simüle ediyor
mse_ois = 5 / (1 + (episodes / 150)**1.8) + np.random.normal(0, 0.05, len(episodes))
mse_wis = 2.5 / (1 + (episodes / 50)**1.5) + np.random.normal(0, 0.05, len(episodes))

# Grafik çizimi
plt.figure(figsize=(8, 5))
plt.plot(episodes, mse_ois, 'g-', label='Ordinary importance sampling')
plt.plot(episodes, mse_wis, 'r-', label='Weighted importance sampling')
plt.xscale('log')
plt.xlabel('Episodes (log scale)')
plt.ylabel('Mean square error (average over 100 runs)')
plt.title('Ordinary vs. Weighted Importance Sampling - MSE Comparison')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
